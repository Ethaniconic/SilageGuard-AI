This notebook documents the conversion and quantization pipeline:
1. **Sensor Random Forest** -> Portable JSON Decision Tree schema
2. **Vision MobileNetV3** -> INT8 quantized TensorFlow Lite container (`.tflite`)
3. Generating metadata, class label dictionaries, and on-device runtime profiles

```python
import os
import json

models_dir = os.path.join('..', 'mobile', 'assets', 'models')
print('Mobile Model Artifacts:')
for f in os.listdir(models_dir):
    p = os.path.join(models_dir, f)
    print(f' - {f}: {os.path.getsize(p)/1024:.1f} KB')
```

## Verification of Sensor Model JSON Schema

```python
rf_json_path = os.path.join(models_dir, 'sensor_rf_model.json')
with open(rf_json_path) as f:
    forest = json.load(f)
print('Model Type:', forest['model_type'])
print('Number of Trees:', forest['n_estimators'])
print('Classes:', forest['classes'])
print('Feature Names:', forest['feature_names'])
```

## Verification of TFLite Model Metadata

```python
meta_path = os.path.join(models_dir, 'model_metadata.json')
with open(meta_path) as f:
    meta = json.load(f)
print(json.dumps(meta, indent=2))
```
